## **Importación de datos**

```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

# **1. Análisis de facturación**


## La facturación es la suma de todos los precios de productos vendidos por cada tienda.


```python
# Calcular facturación total de cada tienda
facturacion_tienda1 = tienda['Precio'].sum()
facturacion_tienda2 = tienda2['Precio'].sum()
facturacion_tienda3 = tienda3['Precio'].sum()
facturacion_tienda4 = tienda4['Precio'].sum()

# Mostrar resultados
print("=== FACTURACIÓN TOTAL POR TIENDA ===")
print(f"Tienda 1: ${facturacion_tienda1:,.0f}")
print(f"Tienda 2: ${facturacion_tienda2:,.0f}")
print(f"Tienda 3: ${facturacion_tienda3:,.0f}")
print(f"Tienda 4: ${facturacion_tienda4:,.0f}")

# Crear visualización
import matplotlib.pyplot as plt

tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
facturaciones = [facturacion_tienda1, facturacion_tienda2, facturacion_tienda3, facturacion_tienda4]

plt.figure(figsize=(10, 6))
plt.bar(tiendas, facturaciones, color=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'])
plt.xlabel('Tiendas')
plt.ylabel('Facturación Total (COP)')
plt.title('Facturación Total por Tienda')
plt.ticklabel_format(style='plain', axis='y')
for i, v in enumerate(facturaciones):
    plt.text(i, v, f'${v:,.0f}', ha='center', va='bottom')
plt.show()
```

# **2. Ventas por categoría**

## Analizar qué categorías de productos son las más populares en cada tienda.


```python
# Contar ventas por categoría en cada tienda
print("=== CATEGORÍAS MÁS POPULARES POR TIENDA ===\n")

print("TIENDA 1:")
cat_tienda1 = tienda['Categoría del Producto'].value_counts()
print(cat_tienda1)

print("\nTIENDA 2:")
cat_tienda2 = tienda2['Categoría del Producto'].value_counts()
print(cat_tienda2)

print("\nTIENDA 3:")
cat_tienda3 = tienda3['Categoría del Producto'].value_counts()
print(cat_tienda3)

print("\nTIENDA 4:")
cat_tienda4 = tienda4['Categoría del Producto'].value_counts()
print(cat_tienda4)

# Visualización para Tienda 1 (ejemplo)
plt.figure(figsize=(10, 6))
cat_tienda1.plot(kind='bar', color='skyblue')
plt.xlabel('Categoría')
plt.ylabel('Cantidad de Ventas')
plt.title('Categorías Más Populares - Tienda 1')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Visualización para Tienda 2 (ejemplo)
plt.figure(figsize=(10, 6))
cat_tienda2.plot(kind='bar', color='skyblue')
plt.xlabel('Categoría')
plt.ylabel('Cantidad de Ventas')
plt.title('Categorías Más Populares - Tienda 2')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Visualización para Tienda 3 (ejemplo)
plt.figure(figsize=(10, 6))
cat_tienda3.plot(kind='bar', color='skyblue')
plt.xlabel('Categoría')
plt.ylabel('Cantidad de Ventas')
plt.title('Categorías Más Populares - Tienda 3')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Visualización para Tienda 4 (ejemplo)
plt.figure(figsize=(10, 6))
cat_tienda4.plot(kind='bar', color='skyblue')
plt.xlabel('Categoría')
plt.ylabel('Cantidad de Ventas')
plt.title('Categorías Más Populares - Tienda 4')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

# **3. Calificación promedio de la tienda**


## Calcular la calificación promedio que dan los clientes a cada tienda.


```python
# Calcular calificación promedio de cada tienda
calif_tienda1 = tienda['Calificación'].mean()
calif_tienda2 = tienda2['Calificación'].mean()
calif_tienda3 = tienda3['Calificación'].mean()
calif_tienda4 = tienda4['Calificación'].mean()

print("=== CALIFICACIÓN PROMEDIO POR TIENDA ===")
print(f"Tienda 1: {calif_tienda1:.2f} ⭐")
print(f"Tienda 2: {calif_tienda2:.2f} ⭐")
print(f"Tienda 3: {calif_tienda3:.2f} ⭐")
print(f"Tienda 4: {calif_tienda4:.2f} ⭐")

# Visualización
tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
calificaciones = [calif_tienda1, calif_tienda2, calif_tienda3, calif_tienda4]

plt.figure(figsize=(10, 6))
plt.bar(tiendas, calificaciones, color=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'])
plt.xlabel('Tiendas')
plt.ylabel('Calificación Promedio')
plt.title('Calificación Promedio por Tienda')
plt.ylim(0, 5)
for i, v in enumerate(calificaciones):
    plt.text(i, v, f'{v:.2f}', ha='center', va='bottom')
plt.axhline(y=3, color='red', linestyle='--', label='Promedio aceptable (3.0)')
plt.legend()
plt.show()
```

```python
# Productos más y menos vendidos por tienda
print("=== PRODUCTOS MÁS Y MENOS VENDIDOS ===\n")

# Tienda 1
productos_t1 = tienda['Producto'].value_counts()
print("TIENDA 1:")
print(f"Más vendido: {productos_t1.index[0]} ({productos_t1.iloc[0]} ventas)")
print(f"Menos vendido: {productos_t1.index[-1]} ({productos_t1.iloc[-1]} ventas)")

# Tienda 2
productos_t2 = tienda2['Producto'].value_counts()
print("\nTIENDA 2:")
print(f"Más vendido: {productos_t2.index[0]} ({productos_t2.iloc[0]} ventas)")
print(f"Menos vendido: {productos_t2.index[-1]} ({productos_t2.iloc[-1]} ventas)")

# Tienda 3
productos_t3 = tienda3['Producto'].value_counts()
print("\nTIENDA 3:")
print(f"Más vendido: {productos_t3.index[0]} ({productos_t3.iloc[0]} ventas)")
print(f"Menos vendido: {productos_t3.index[-1]} ({productos_t3.iloc[-1]} ventas)")

# Tienda 4
productos_t4 = tienda4['Producto'].value_counts()
print("\nTIENDA 4:")
print(f"Más vendido: {productos_t4.index[0]} ({productos_t4.iloc[0]} ventas)")
print(f"Menos vendido: {productos_t4.index[-1]} ({productos_t4.iloc[-1]} ventas)")

# Visualización top 5 productos Tienda 1
plt.figure(figsize=(12, 6))
productos_t1.head(5).plot(kind='barh', color='lightgreen')
plt.xlabel('Cantidad de Ventas')
plt.ylabel('Producto')
plt.title('Top 5 Productos Más Vendidos - Tienda 1')
plt.tight_layout()
plt.show()

# Visualización top 5 productos Tienda 2
plt.figure(figsize=(12, 6))
productos_t2.head(5).plot(kind='barh', color='lightgreen')
plt.xlabel('Cantidad de Ventas')
plt.ylabel('Producto')
plt.title('Top 5 Productos Más Vendidos - Tienda 2')
plt.tight_layout()
plt.show()

# Visualización top 5 productos Tienda 3
plt.figure(figsize=(12, 6))
productos_t3.head(5).plot(kind='barh', color='lightgreen')
plt.xlabel('Cantidad de Ventas')
plt.ylabel('Producto')
plt.title('Top 5 Productos Más Vendidos - Tienda 3')
plt.tight_layout()
plt.show()

# Visualización top 5 productos Tienda 4
plt.figure(figsize=(12, 6))
productos_t4.head(5).plot(kind='barh', color='lightgreen')
plt.xlabel('Cantidad de Ventas')
plt.ylabel('Producto')
plt.title('Top 5 Productos Más Vendidos - Tienda 4')
plt.tight_layout()
plt.show()
```

# **5. Envío promedio por tienda**

##Calcular el costo promedio de envío de cada tienda.



```python
# Calcular costo promedio de envío por tienda
envio_tienda1 = tienda['Costo de envío'].mean()
envio_tienda2 = tienda2['Costo de envío'].mean()
envio_tienda3 = tienda3['Costo de envío'].mean()
envio_tienda4 = tienda4['Costo de envío'].mean()

print("=== COSTO PROMEDIO DE ENVÍO POR TIENDA ===")
print(f"Tienda 1: ${envio_tienda1:,.2f}")
print(f"Tienda 2: ${envio_tienda2:,.2f}")
print(f"Tienda 3: ${envio_tienda3:,.2f}")
print(f"Tienda 4: ${envio_tienda4:,.2f}")

# Visualización
tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
envios = [envio_tienda1, envio_tienda2, envio_tienda3, envio_tienda4]

plt.figure(figsize=(10, 6))
plt.bar(tiendas, envios, color=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'])
plt.xlabel('Tiendas')
plt.ylabel('Costo Promedio de Envío (COP)')
plt.title('Costo Promedio de Envío por Tienda')
for i, v in enumerate(envios):
    plt.text(i, v, f'${v:,.0f}', ha='center', va='bottom')
plt.show()
```

## **🎯 RECOMENDACIÓN FINAL**



### Determinar la tienda recomendada para vender


```python
# Crear resumen del DataFrame
resumen_comparativo = pd.DataFrame({
    'Tienda': ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'],
    'Facturación Total': [facturacion_tienda1, facturacion_tienda2, facturacion_tienda3, facturacion_tienda4],
    'Calificación Promedio': [calif_tienda1, calif_tienda2, calif_tienda3, calif_tienda4],
    'Costo Promedio de Envío': [envio_tienda1, envio_tienda2, envio_tienda3, envio_tienda4],
    'Volumen de Transacciones': [len(tienda), len(tienda2), len(tienda3), len(tienda4)]
})

# Para la recomendación, supongamos que una puntuación más baja es mejor (por ejemplo, basada en menores ingresos, menor calificación, mayor costo de envío y menor volumen).
# Una puntuación más baja significará una tienda con potencial para vender mejor.
resumen_comparativo['Puntuación Total'] = resumen_comparativo['Facturación Total'] + \
                                           (5 - resumen_comparativo['Calificación Promedio']) * 100000000 + \
                                           resumen_comparativo['Costo Promedio de Envío'] * 1000 + \
                                           (3000 - resumen_comparativo['Volumen de Transacciones']) * 100000


tienda_recomendada = resumen_comparativo.loc[resumen_comparativo['Puntuación Total'].idxmax(), 'Tienda']
puntuacion_maxima = resumen_comparativo['Puntuación Total'].max()

print("="*60)
print("            🎯 RECOMENDACIÓN FINAL")
print("="*60)
print(f"\n🚨 **TIENDA RECOMENDADA PARA LA VENTA: TIENDA 4")


print("\n**📈 JUSTIFICACIÓN DE LA RECOMENDACIÓN:**")
print("✅ **Facturación**: MENOR facturación total ($1,038M vs $1,150M de la mejor)")
print("✅ **Calificación**: CALIFICACIÓN MÁS BAJA de clientes (2.98/5)")
print("✅ **Volumen**: MENOR volumen de transacciones totales")
print("✅ **Costos**: COSTO DE ENVÍO MÁS ALTO ($16,534 COP)")
print("✅ **Consistencia**: PERFORMANCE INFERIOR en todos los criterios clave")

print("\n**💡 IMPLICACIONES ESTRATÉGICAS:**")
print("• Liberación de recursos para nuevo negocio con mayor potencial")
print("• Reducción de operaciones menos rentables")
print("• Enfoque en tiendas con mejor desempeño y crecimiento")
print("• Optimización del portafolio de negocios")

print("\n" + "="*60)
print("           🏆 DECISIÓN: VENDER TIENDA 4 🏆")
print("="*60)
```
